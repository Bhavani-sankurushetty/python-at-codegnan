```python
# FizzBuzz problem
n=int(input())
if n%3==0 and n%5==0:
    print("FizzBuzz")
elif n%3==0:
    print("Fizz")
elif n%5==0:
    print("Buzz")
else:
    print ("Zero")
```

```python
# +ve, -ve and zero
n=int(input())
if n > 0:
    print("+ve")
elif n < 0:
    print("-ve")
else:
    print("zero")
```

```python
num = int(input())
if num >=0 and num % 2 == 0:
    print(f"{num} is +ve Even")
elif num>=0:
    print(f"{num} is +ve odd")
elif num < 0 and num % 2 == 0:
    print(f"{num} is -ve Even")
elif num < 0:
    print(f"{num} is -ve odd")
else:
    print("zero")
```

```python
# fine the smallest number among the three numbers
n=list(map(int,input().split()))
print(min(n))
```

```python
# smallest number using loop function
n=list(map(int,input().split()))
mini=n[0]
for i in n:
    if i < mini:
        mini = i
print (mini)
```

```python
# using nested for loop print 
n = int(input())
if n >= 0:
    if n % 2==0:
        print ("+ve even")
    else:
        print("+ve odd")
else:
    if n % 2==0:
        print ("-ve even")
    else:
        print("-ve odd")
```
