```python
#method overloading
class math:
    def add(x,y):
        return x+y
    def add(x,y,z):
        return x+y+z
    def add(a,b,c,d):
        return a+b+c+d
print(math.add(1,2,3,4))
print(math.add(1,2,3))
print(math.add(1,2))
```

```python
#python directing doesnot support method overloading,but we can achive through the default arguments or arbitary arguments
class math:
    def add(a,b,c=0,d=0):
        return a+b+c+d
print(math.add(1,2,3,4))
print(math.add(1,2,3))
print(math.add(1,2))
```
