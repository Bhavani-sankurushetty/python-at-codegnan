```python
# reading integer from user
# integer to float, string, bool
n = int(input())
print(type(n), n)
print(float(n))
print(str(n))
print(bool(n))
```

```python
# read float value from user
# float to int, string, bool
val = float(input())
print(type(val), val)
print(int(val))
print(str(val)[:4])
print(bool(val))
```

```python
# reading complex value
real = int(input())
img = int(input())
c = complex(real, img)
print(type(c), c)
```

```python
# reading string value from user
s = input() # "123"
print(type(s), s)
print(int(s))
print(float(s))
print(list(s)) # it returns list of characters
print(tuple(s)) # it returns tuple of characters
print(set(s)) # it returns set of characters
```
