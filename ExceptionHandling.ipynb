```python
# Exception handling block
# without exception
try:
    a=int(input())
    b=int(input())
    c=a+b
    print(c)
except:
    print("Something Wrong in try block")
else:
    print("There is no exception")
finally:
    print("Program done")
```

```python
# Exception handling block
# with exception
try:
    a=int(input())
    b=int(input())
    c=a+b
    print(c)
except:
    print("Something Wrong in try block")
else:
    print("There is no exception")
finally:
    print("Program done")
```

```python
# Exception Handling Block
# Exception with an exception class
try:
    a=10
    b=0
    c=a/b
    print(c)
except Exception as e:
    print(f"Something Wrong in try block:{e}")
```

```python
# Exception Handling Block
# define multiple exception with except block
try:
    a=10
    b=0
    c=a/b
    print(c)
except ZeroDivisionError as e:
    print(f"Zero division not possible:{e}")
except IndexError as e:
    print(f"we can't access element at this position: {e}")
except Exception as e:
    print(f"Something Wrong in try block:{e}")
```

```python
# Exception Handling Block
# define multiple exception with except block
try:
    arr = [1,2,3,4]
    c=arr[10]
    print(c)
except ZeroDivisionError as e:
    print(f"Zero division not possible:{e}")
except IndexError as e:
    print(f"we can't access element at this position: {e}")
except Exception as e:
    print(f"Something Wrong in try block:{e}")
```
