### set methods
- Adding Element to set
  - add(ele): it adds element to set if element not exists in set, otherwise nothing
  - update(iterable): it adds all iterable elements to set
- Remove elements from set
  - remove(ele): It removes the element from set if element exists in set otherwise `key Error`
  - discard(ele): it removes the element from set if element exists in set otherwise nothing
  - pop(): it removes and returns the element from set
- set operations
  - union(another_sets): it returns all the common elements from all sets
  - intersection(another_sets): it returns the common elements from all the sets
  - difference(another_sets): it returns the set elements those are not lies in another_sets
  - symmetric_difference(another_sets): it returns the non-common elements from all sets

```python
s = {1,2,3,-5}
s.add(4)
print(s)
s.add(1) # 1 dosent 
print(s)
```

```python
s1 = {1,2,5,7}
s2 = {3,4,5,6,7}
s3 = {7,9}
print(s1.union(s2))
print(s1.union(s2,s3))
```

```python
s1 = {1,2,5,7}
s2 = {3,4,5,6,7}
print(s1.intersection(s2))
print(s1.difference(s2))
print(s1.symmetric_difference(s2))
```

```python
s1 = {1,2,5,7}
s2 = {3,4,5,6,7}
print(s1 | s2) # |- union
print(s1 & s2) # &- intersection
print(s1 - s2) # - for difference
print(s1 ^ s2)# ^ for symmetric difference
```

## set checking methods
- issubset(another_set)
- issuperset(another_set)
- isdisjoint(another_set)

```python
print({1,2,3}.issubset({1,2,3,4,4}))
print({1,2,3,4}.issuperset({1,2}))
print(set().issubset({1}))  
print({1,2}.isdisjoint({4,5}))
```
