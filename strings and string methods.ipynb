#### String Methods
- Case Conversion Methods
  - upper() : it converts entire string into uppercase
  - Lower() : it converts entire string into lowercase 
  - title() : it converts every word first character into uppercase and remaining into lowercase
  - swapcase() : it converts lower into upper and upper to lowercase
  - capitalize() : it converts first character of string into uppercase and remaining into lowercase

```python
string = "The ocean is blue and sky is BLUE"
lower = string.lower()
print(lower)
print(string.upper())
print(string.title())
print(string.swapcase())
print(string.capitalize())
```

```python
# White Space remove methods
# strip() : it removes the white space to both sides of string
# lstrip() : it removes the white space from the left side of string (leading white space)
# rstrip() : it removes the white space from the right side of string (Leading white space)
string= "      Codegnan       "
print(string)
print(string.strip()) 
print(string.lstrip())
print(string.rstrip())
```

```python
string= "      Code   gnan       "
print(string)
print(string.strip()) 
print(string.lstrip())
print(string.rstrip())
```

```python
string= "@@@@@@@ Code @ gnan@@@@@@"
print(string)
print(string.strip("@")) 
print(string.lstrip("@"))
print(string.rstrip())
```

```python
# split(): It returns the list of substring and by default splits at white space
string = "The ocean is blue and sky is BLUE"
print(string.split())
print(string.split('blue'))
```

# Indexing finding Methods
 - index(sub_string) : it returns the first occurance index of sub-string if exists, otherwise "valueError"
 - find(sub_string) : it returns the first occurance index of sub-string if exists, otherwise "-1"

```python
string = "The ocean is blue and sky is BLUE"
print(string.index("o"))
print(string.index('is'))
print(string.index('is', 14)) # it starts checking from the 14th index
print(string.index('Z')) # error
```

```python
string = "The ocean is blue and sky is BLUE"
print(string.find("o"))
print(string.find('is'))
print(string.find('is', 14)) # it starts checking from the 14th index
print(string.find('Z')) # it returns -1
```

# replace(old ,new):

#### string checking Methods
- islower() : it returns true when entire string is in lower case
- isupper() : it returns true when entire string is in upper case
- istitle() : it returns true when string is titled format
- isalpha() : it returns true when string contains alphabets(i.e lowercase and uppercase)
- isalnum() : it returns true when string contains either alphabets or digits
- isdigit() : it returns true when string contains only digits 
- startswith() : it returns true when string startswith the sub_string
- endswith() : it returns true when string endswith the sub_string

```python
print("hello".islower())
print("Hello".islower())
print("HELLO".isupper())
print("1234".isalpha())
print("abcDE".isalpha())
print("1234".isalnum())
print("1343abc".isalnum())
print("Hello World".startswith('H'))
print("Hello World".endswith('H'))
```
