```python
class Account:
    def __init__(self,username,balance,password):
        self.username=username
        self.balance=balance
        self.password=password
    def withdraw(self,amount):
        if self.balance >= amount:
            self.balance -= amount
            return f"{amount} withdrawal successfully"
        return "Insufficient Balance"
    def deposit(self, amount):
        self.balance += amount
        return f"{amount} Deposit successfully and current balance is {self.balance}"
user = Account("bhavani", 5000, "bhavani@123")
print(user.deposit(1000))
user.balance = 100000
print(user.withdraw(50000))
```
