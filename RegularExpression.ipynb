```python
import re
```

```python
# match(pat, string)
string = "The sky is blue and Ocean is blue"
res = re.match("[A-Z]", string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # it returns matched patterns val
```

```python
# match(pat, string)
string = "The sky is blue and Ocean is blue"
res = re.match("[A-Za-z]+", string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # it returns matched patterns val
```

```python
# match(pat, string)
try:
    string = "The sky is blue and Ocean is blue"
    res = re.match("[z]+", string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched patterns val
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
#search(pat, string)
# search 5 length word
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.search("[a-zA-Z]{5}+", string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched patterns val
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# get the words from string which length is more than 3
#findall(pat, string)
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.findall("[a-zA-Z]{3,}", string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched patterns val
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# get the words from string which length is more than 3
#findall(pat, string)
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.findall("[a-zA-Z]{3,}", string)
    print(res)
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# validate mobile number
# All are digits, length of number is 10, strats with 6, 7, 8, 9, ramaining are 0-9
import re
try:
    mobile = input()
    res = re.findall("^[6-9][0-9]{9}", mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# PAN Number validation
# it allows upper and digits, length is 10, first 5 are upper, next 4 are digits, and last one ends with upper
import re
try:
    pan = input()
    res = re.findall("^[A-Z]{5}[0-9]{4}[A-Z]$", pan)
    print(res)
    if res:
        print(f"{pan} is valid")
    else:
        print(f"{pan} is invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```
