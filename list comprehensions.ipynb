```python
# list comprehension
# syntav: [exp for iterator in iterable if condition]
# return 1-n numbers as list using list comprehension
n = int(input())
numbers = [i for i in range(1, n + 1)]
print(numbers)
```

```python
n = int(input())
res = [2 * i for i in range(1, n + 1)]
print(res)
```

```python
n = int(input())
res = [2 * i for i in range(n)]
print(res)
```

```python
n = int(input())
res = [i for i in range(0, 2*n, 2)]
print(res)
```

```python
# return list even or odd, for list of integers
arr = [20,34,33,55,23,54]
res = ["Odd" if x&1 else "Even" for x in arr]
print(res)
```

```python
arr = [20,34,33,23,54]
res = [x+0.5 if x&1 else x+1 for x in arr]
print(res)
```

```python
# find one dice combinations
print([x for x in range(1,7)])
```

```python
# find two dices combinations
print([(i, j) for i in range(1,7) for j in range(1,7)])
```

```python
# find two dices combinations if any dice face value is 4
print([(i, j) for i in range(1,7) for j in range(1,7) if i==4 or j==4])
```
