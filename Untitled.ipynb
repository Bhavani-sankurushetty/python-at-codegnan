```python
# find factors of a numbers
num = int(input())
factors = []
for i in range(1, num+1):
    if num%i == 0:
        factors.append(i)
print(factors)
if len (factors) == 2:
    print(f"{num} is prime")
else:
    print(f"{num} is not prime")
```

```python
num = int(input())
for i in range(2, num//2+1):
    if num%i == 0:
        print(f"{num} is not prime")
        break
else:
    print(f"{num} is prime")
```

```python
# find reverse of a string
string = input()
rev = ""
for char in string:
    rev = char + rev
print(rev)
if rev == string:
    print(f"{string} is palindrome")
else:
    print(f"{string} is not palindrome")
```
