```python
# Dictionary
details = { 'name':'bhavani','age':21, 'course':'python'}
print(details) # dictionary is ordered collection
```

```python
# dictionary cant allow list as key
# if dictionary contains multiple items with key, in that case it updates first item key value with last defined item value
details = { 'name':'bhavani','age':21, 'course':'python','marks':[45,96,48]}
print(details) # dictionary is ordered collection
```

```python
# access values by using keys
details = { 'name':'bhavani','age':21, 'course':'python','marks':[45,96,48], 'age':50}
print(details['course'])
print(details['age'])
# updating the Dictionary value
details['age'] += 1
details['course'] = 'java' # it updates the value for 'course' key 
print(details)
# item assignment
details['location'] = "India" # it adds new item to dictionary
print(details)
```

#### Dictionary Methods
- Get value
   - get(key, default_val): it returns the value if key exists in dictionary otherwise `default_Value`
- Update values and add items to dictionary
   - update(new_dictionary): Dictionary Updates with New_dictionary Items
- Remove items from dictionary
   - pop(key): it returns value and remove item if key is exists in dictionary otherwise `KeyError`
   - popitem(): it returns tuple of key value pair and removes the last item from dictionary
- Get Keys, value and Items from dictionary
   - Key(): it returns list of dictionary Keys
   - values(): it returns list of disctionary values
   - items(): it returns list of tuple key_value pairs

```python
# using get
details = { 'name':'bhavani','age':21, 'course':'python' }
d = {'marks': [45,96,48], 'age':50}
details.update(d)
print(details)
```

```python
details = { 'name':'bhavani','age':21, 'course':'python','marks':[45,96,48], 'age':50}
print(details)
print(details.pop('age'))
print(details)
print(details.popitem())
print(details)
# print(details.pop('precentage')) # keyError: percentage
```
