```python
# local scope
# we can access the local scope variable within the scope, we cant access in global scope
def fun(x):
    x = x + 1 # x is the local scope
    return x 
a = 10
 # res = fun(a)
print(fun(a))
```

```python
# enclosing scope
def outer(x):
    y = 10 # enclosing scope
    def inner():
        a = 10 # local scope
        z = y+a
        print(f"this is local scope:{z}")
    inner()
    y = y + x
    print(f"this is enclosing scope:{y}")
n = 5
outer(n)
```

```python
def outer(x):
    x = 10 # enclosing scope
    def inner():
        x = 20 # local scope
        z = x+x
        print(f"this is local scope:{z}")
    inner()
    y = x + x
    print(f"this is enclosing scope:{y}")
n = 5
outer(n)
```

```python
# global scope
x = 10 # global scope
def outer():
    y = 20 # enclosing scope
    def inner():
        z = 30 # local scope
        res = z+y+x
        print(f"this is local scope:{res}")
    res = y + x # if y=y+x now 20=20+10=30 and the resulting in 70 for the inner()
    inner()
    print(f"this is enclosing scope:{res}")
outer()
```

```python
#  update the global scope variable inside the  local scope by using 'global' keyword
x = 10 # global scope
def outer():
    global x 
    x = x+1
    print("this is local scope val:",x)
outer()
print("this is global scope val:", x)
```

```python
# trying to update enclosing scope variable in local scope using 'nonlocal' keywords
x = 10 # global scope
def outer():
    y = 20 # enclosing scope
    def inner():
        nonlocal y
        z = 30 # local scope
        y = y+1
        print(f"this is local scope:{y}")
    y = y + x #30
    inner()
    y = y + x #30
    print(f"this is enclosing scope:{y}")
outer()
```
