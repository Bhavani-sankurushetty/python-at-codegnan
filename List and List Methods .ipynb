#### List Methods
- Element Adding Methods
  - append(item) :  It adds item at end of the list
  - expand(iterable) : it adds all iterable elements at end of the list
  - insert(pos, item) : this method adds the item at specified position

```python
lst = [1,2,3,4]
print(lst.append(10))
print(lst)
lst.append('code')
print(lst)
lst.append(['10,20,30'])
print(lst)
print(lst[6])
```

```python
lst = [1,2,3,4]
print(lst.append(10))
print(lst)
lst.append('code')
print(lst)
lst.append(['10,20,30'])
print(lst)
# print(lst[6])
lst.extend([100,200,300])
print(lst)
lst.insert(1,2000)
print(lst)
```

## Element remove methods
  - remove(val) : it removes the first occurence of value if value exists in list,otherwise it raises 'ValueError'
  - pop() : it remove and return last element of list 
  - pop(pos) : it removes and returns the specified Position element form list
  - clear() : it removes all elements from list 
  - del : it is a python keyword used to delete values

```python
lst = [10,20,30,20,40]
lst.remove(20)
print(lst)
# list.remove(100) # ValueError: list.remove(x): x not in list
val = lst.pop()
print(val, lst)
val2 = lst.pop() # it removes the ist position element
print(val2, lst)
# print(lst.pop(10))# IndexError:pop index out of range
del lst[0]
print(lst)
lst.clear()
print(lst)
```

```python
# index(val): it returns the index of first occurence of value if value exists in list otherwise valueError
# count(val): it returns the total frequence of value
lst = [11,2,12,4,14,2,2,2]
ind = lst.index(2)
print(ind)
print(lst.count(2))
print(lst.count(100))
print(lst.index(100))
```

```python
# sort(reverse =False): It sorts the all list elements in ascending order, If reverse value is true it sorts in descending order.
lst = [5,43,63,73,-2,4,-6]
lst.sort(reverse=True)
print(lst)
```
