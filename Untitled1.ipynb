```python
# types of arguments
# 1. positional arguments : the arguments are assigns to parameters as per the positional order 
def sub(x:int, y:int)->int:
    c = x-y
    return c 
# function call block
a,b = 10,20
res = sum(a,b)
print(res)
print(sub(b,a))
```

```python
# types of arguments
# 2. keyword arguments : here arguments are passed along with the parameters
def sub(x:int, y:int)->int:
    c = x-y
    return c 
# function call block
a,b = 10,20
print(sub(x = a, y = b))
print(sub(y = b, x = a))
```

```python
# default arguments: the parameters define with some default value, and it assigns the argument value if you pass the values for default parameters otherwise, it will assign the default value
def add(a,b,c = 0, d = 0):
    res = a+b+c+d
    return res
print(add(10,20))
print(add(10,20,5))
print(add(10,20,2,1))
```

```python
# pass by value
def fun(x):
    print(x)
    print(f"Before update Address of x : {id(x)}")
    print(x)
    x = x + 1
    print(x)
    print(f"After update the x value, address of x :{id(x)}")
a = 5
print(f" Before function execution address of a: (id(a) and value: {a}")
print(fun(a))
print(f" After function execution address of a: (id(a) and value: {a}")
```

```python
def fun(x):

    print(f"Before update Address of x : {id(x)} and value: {x}")
    x = x + 1
    print(f"After update the x value, address of x :{id(x)} and value: {x}")
a = 5
print(f" Before function execution address of a: (id(a) and value: {a}")
print(fun(a))
print(f" After function execution address of a: (id(a) and value: {a}")
```

```python
# pass by reference
def fun(x):
    print(f"Before update Address of x : {id(x)} and value: {x}")
    x.append(100)
    print(f"After update the x value, address of x :{id(x)} and value: {x}")
a = [1,2,3]
print(f" Before function execution address of a: (id(a) and value: {a}")
print(fun(a))
print(f" After function execution address of a: (id(a) and value: {a}")
```
