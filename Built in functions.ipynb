```python
# abs() : it returns the absolute value
print(abs(20-10))
print(abs(10-20))
```

```python
# round(val, pos): this function rounds the float value to particular positions
print(round(2.2589, 2))
print(round(2.2539, 2))
print(round(2.2589))
print(int(2.2539))
```

```python
# ord(char) : it returns ASCII value of character
# chr(ascii_val) : it returns the character for specified ASCII value
print(ord("A"))
print(chr(65))
```

```python
# sorted(iterable, reverse=False, key=function): it  returns new isorted iterable
# sort the fruits based in the length of fruit name
fruits = ['Apple', 'banana', 'kiwi', 'Mango', 'Watermelon']
res = sorted(fruits)
print(res)
print(sorted(fruits, reverse = True))
print(sorted(fruits, reverse = True, key=len)) # it sorts the list based on length of each item in a length
```

```python
# map(function, iterable)
# find the square of all list of integers using map function
def square(x):
    return x**2
arr = [1,2,3,4,5]
res = list(map(square, arr))
print(res)
# filter(function, iterable)
# Find even numbers from list of numbers
lst = [1,2,3,4,5,6]
```

```python
# filter(function, iterable) : it filters the iterable, if the function returns non-None value, those values will added to result 
# Find even numbers from list of numbers
def check_even(x):
    if x % 2 ==0:
        return x
lst = [1,2,3,4,5,6]
res = list(filter(check_even, lst))
print(res)
```

```python
# find the 3's multipliers from list of integers
arr = [1,24,54,22,44,51,33,15]
def three_multipliers(x):
    if x%3 == 0:
        return 1
res = list(filter(three_multipliers, arr))
print(res)
```

```python
# zip(iterable1, iterable2, ...): it returns all iterable as single iterable of tuples
roll_no =[101,102,103,104,105]
names = ['Ravi', 'Raju', 'siva', 'Rani', 'Gopi']
percentage = [90,67,96.8,67.65,99.99999]
print(list(zip(roll_no, names, percentage)))
```

```python
# conversion of decimal to binary
num = int(input())
res = ""
while num > 0:
    bit = num % 2
    num = num // 2
    res = str(bit) + res
print(res)
```

```python
# number conversions
print(bin(10)) # string type of binary value
print(oct(10)[2:])
print(oct(10))
print(hex(11).upper()[2:])
```

```python
# any() : this function returns true when any one iterable items return true
# all() : this function returns true when all iterable items return true
arr = [1,2,3,4,0]
print(all(arr))
print(any(arr))
```

```python
percentage = [34,35,54,65,67]
lst = []
for num in percentage:
    if num < 35:
        lst.append(False)
    else:
        lst.append(True)
print(lst)
if all(lst):
        print("all students passed")
else:
        print("all students not pass")
```
