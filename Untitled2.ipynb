```python
# generator defination
def gen(x):
    x = x+1
    yield x
    x = x+2
    yield x
    x = x+3
    yield x
res = gen(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
print(next(res))
```

```python
# generator defination
def gen(x):
    x = x+1
    yield x
    x = x+2
    yield x
    x = x+3
    yield x
res = gen(5)
for i in res:
    print(i)
```

```python
# 1 to n by using functionsand generators
def fun(x):
    res = []
    for i in range(1, x+1):
        res.append(i)
    return res
print(fun(10))
    
```

```python
def gen(x):
    res = []
    for i in range(1, x+1):
        yield i
res = gen(10)
for i in res:
    print(i)
```
