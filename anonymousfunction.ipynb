```python
# fine square of a number using lambda
add = lambda x, y : x + y
print(add(3,4))
```

```python
# find square of all list of integers using lambda and map()
arr = [1,2,3,4,5]
res = list(map(lambda x : x**2, arr))
print(res)
```

```python
# find even numbers in list of integers using lambda and filter()
arr = [1,2,3,4,5]
res = list(filter(lambda x : x%2==0, arr))
print(res)
```

```python
# single line if-else syntax
#True_exp if condition else False_exp
# find whether the number is even or odd using lambda
n = int(input())
even_or_odd = lambda x :f"{x} - odd" if x%2 else f"{x} - even"
print(even_or_odd(n))
```
